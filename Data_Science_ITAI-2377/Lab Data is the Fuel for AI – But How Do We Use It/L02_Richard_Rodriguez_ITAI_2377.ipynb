 Part 4: Understanding Image Data (Computer Vision)
Unlike structured or text data, image data requires different processing.

Load CIFAR-10 Using PyTorch (TorchVision)

```python
import torchvision
import torchvision.transforms as transforms
# Load CIFAR-10 dataset from TorchVision
dataset = torchvision.datasets.CIFAR10(root="./data", train=True, download=True, transform=transforms.ToTensor())
# Extract a subset of 10 images per class
class_indices = {i: [] for i in range(10)}

for idx, (_, label) in enumerate(dataset):
    if len(class_indices[label]) < 10:
        class_indices[label].append(idx)

# ✨ TIP: What function in PyTorch DataLoader allows sampling specific indices?
```

#Discussion Questions:#
How does AI process structured, image, and text data differently?

AI treats each of these data types differently because each one carries information in a different form.
Structured data is stored in rows and columns with clearly defined values either numerical or in categories. This is normally provided to AI in tables and CSV spreadsheets.
Images are displayed to AI as pixels in grids that it then uses convolutional networks or vision transformers to understand and find the patterns in to identify the images.
Text data is unstructured and comes from books, newspapers, and various forms of written works. AI stores this data as tokens and learns the relationships between the words to make predictions.
